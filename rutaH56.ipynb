```python
# BASE DE CONOCIMIENTO (Ruta H56: Portal El Dorado - Portal Tunal)
#Diccionario con las rutas a utilizar BD llave - valor
rutas = {
    "PORTAL EL DORADO": ["NORMANDIA"],
    "NORMANDIA": ["PORTAL EL DORADO", "CAN"],
    "CAN": ["NORMANDIA", "GOBERNACION"],
    "GOBERNACION": ["CAN", "CAD"],
    "CAD": ["GOBERNACION", "GUATOQUE VERAGUAS"],
    "GUATOQUE VERAGUAS": ["CAD", "TYGUA SAN JOSE"],
    "TYGUA SAN JOSE": ["GUATOQUE VERAGUAS", "NARIÑO"],
    "NARIÑO": ["TYGUA SAN JOSE", "OLAYA"],
    "OLAYA": ["NARIÑO", "QUIROGA"],
    "QUIROGA": ["OLAYA", "PARQUE"],
    "PARQUE": ["QUIROGA", "BIBLIOTECA"],
    "BIBLIOTECA": ["PARQUE", "PORTAL TUNAL"],
    "PORTAL TUNAL": ["BIBLIOTECA"]
}

# REGLA PARA BUSCAR UNA RUTA
#Recibe el origen
def buscar_ruta(origen, destino, ruta=None):
    if ruta is None:
        ruta = []

    ruta = ruta + [origen]

    # Si llegamos al destino
    if origen == destino:
        return ruta

    # Revisa las estaciones conectadas
    for estacion in rutas[origen]:
        if estacion not in ruta:
            nueva_ruta = buscar_ruta(estacion, destino, ruta)
            if nueva_ruta:
                return nueva_ruta

    return None

# PROGRAMA PRINCIPAL
print("=========================================")
print("  SISTEMA INTELIGENTE DE RUTAS H56      ")
print("=========================================")

print("\nEstaciones disponibles en la ruta:")
for estacion in rutas.keys():
    print(f" • {estacion}")

# Permite ingresar por teclado la ruta seleccionada
origen = input("\nIngrese la estación de origen: ").strip().upper()
destino = input("Ingrese la estación de destino: ").strip().upper()

# Validación
if origen not in rutas or destino not in rutas:
    print("\nLa estación ingresada no existe en el sistema.")
else:
    resultado = buscar_ruta(origen, destino)

    if resultado:
        print("\nRuta encontrada:")
        print(" ➔ ".join(resultado))
        print(f"\nNúmero total de estaciones recorridas: {len(resultado)}")
    else:
        print("\nNo existe una ruta disponible entre las estaciones.")
```
